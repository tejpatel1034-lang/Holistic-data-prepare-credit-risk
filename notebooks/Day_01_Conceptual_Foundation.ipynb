## Part A – Conceptual Foundation

### Task 1.1 – What is Data Analysis?

**Definition:** Data analysis is the systematic process of inspecting, cleaning, transforming and interpreting data to discover useful information, identify patterns, answer questions and support decisions.

**In simple terms:** Data analysis means converting raw data into meaningful information.

**Fintech example:** A bank can analyse customers' income, credit scores, loan amounts and repayment histories to understand repayment behaviour and identify customers who may default.

**Typical stages:**
1. Define the question or business objective.
2. Collect relevant data.
3. Inspect and clean the data.
4. Explore patterns using statistics and visualisations.
5. Interpret findings.
6. Communicate results and recommendations.

**Data analysis vs. data science:** Data analysis primarily focuses on understanding data and answering questions. Data science is broader and may also include experimentation, predictive modelling, machine learning and deployment.

### Task 1.2 – How to Plan a Data Science Project

| Step | Activity | Customer credit risk example |
|---|---|---|
| 1 | Business understanding | Reduce loan losses caused by defaults |
| 2 | Define success | Identify customers with a higher default risk |
| 3 | Data understanding | Review customer, loan and repayment data |
| 4 | Data acquisition | Collect CSV, JSON, SQL and API data |
| 5 | Data preparation | Clean missing values, duplicates and outliers |
| 6 | Feature engineering | Create debt-to-income and repayment features |
| 7 | Modelling | Train a classification model |
| 8 | Evaluation | Evaluate precision, recall, F1-score and ROC-AUC |
| 9 | Deployment and monitoring | Integrate the model and monitor performance |


### Task 1.3 – How to Frame a Machine Learning Problem

Problem framing converts a business question into a clearly defined machine-learning task.

**Business question:** Is a customer likely to default on a loan?

**Machine-learning formulation:**
- **Problem type:** Supervised learning – binary classification.
- **Input features (X):** Age, annual income, loan amount, credit score, repayment history, transaction count and other permitted customer attributes.
- **Target (y):** `default_flag`.
- **Class 0:** No default.
- **Class 1:** Default.
- **Prediction:** The model estimates the probability or class of loan default.

**Example:** If the model predicts a default probability of 0.82, it estimates an 82% probability under the model's assumptions. A lender must choose a decision threshold based on business costs, risk policy and fairness requirements.

**Suitable evaluation metrics:**
- Recall for class 1: how many actual defaulters are identified.
- Precision for class 1: how many customers predicted as defaulters actually default.
- F1-score: harmonic mean of precision and recall.
- ROC-AUC: ability to rank positive cases above negative cases.

Accuracy alone can be misleading when defaults are rare.

**Important modelling considerations:** Avoid data leakage, use a suitable train/test split, protect customer privacy, assess fairness, and do not treat a model prediction as the only basis for a lending decision.

### Task 2 – Explain Tensors with NumPy Examples

**Definition:** A tensor is a generalised multi-dimensional array. A scalar is a 0-D tensor, a vector is a 1-D tensor, a matrix is a 2-D tensor, and an array with three or more dimensions is a higher-dimensional tensor.

In NumPy, `ndarray` is the main structure used to represent arrays and tensors.

| Tensor order | Common name | Example |
|---|---|---|
| 0-D | Scalar | One credit score |
| 1-D | Vector | Credit scores of several customers |
| 2-D | Matrix | Rows = customers, columns = features |
| 3-D | 3-D tensor | Multiple customer-feature matrices, such as batches |

Important attributes:
- `ndim`: number of dimensions
- `shape`: size along each dimension
- `size`: total number of elements
- `dtype`: data type of elements

#### 2.1 Create tensors of different dimensions

```python
# 0-D tensor: a single scalar value
scalar = np.array(720)

# 1-D tensor: credit scores of three customers
vector = np.array([720, 650, 780])

# 2-D tensor: rows are customers; columns are [age, credit_score]
matrix = np.array([
    [25, 720],
    [41, 650],
    [33, 780]
])

# 3-D tensor: two batches, each containing two customers and two features
tensor_3d = np.array([
    [[25, 720], [41, 650]],
    [[33, 780], [29, 690]]
])

for name, arr in [
    ("Scalar", scalar),
    ("Vector", vector),
    ("Matrix", matrix),
    ("3-D Tensor", tensor_3d)
]:
    print(f"{name}:")
    print(arr)
    print("ndim:", arr.ndim, "| shape:", arr.shape,
          "| size:", arr.size, "| dtype:", arr.dtype)
    print("-" * 55)
```

#### 2.2 Indexing and slicing

```python
print("First customer's credit score:", matrix[0, 1])
print("All customer ages:", matrix[:, 0])
print("All credit scores:", matrix[:, 1])
print("First batch from 3-D tensor:\n", tensor_3d[0])
```

#### 2.3 Tensor arithmetic and broadcasting

```python
scores = np.array([600, 650, 700, 750])
adjustment = 20

print("Original scores:", scores)
print("Scores after adding 20:", scores + adjustment)
print("Scores multiplied by 1.05:", scores * 1.05)

# Broadcasting: subtract the feature-wise mean from every row
customer_features = np.array([
    [25, 720],
    [41, 650],
    [33, 780]
])
feature_means = customer_features.mean(axis=0)

centered_features = customer_features - feature_means
print("Feature means:", feature_means)
print("Centered features:\n", centered_features)
```

#### 2.4 Reshaping and matrix multiplication

```python
values = np.arange(1, 7)
reshaped = values.reshape(2, 3)

print("Original:", values)
print("Reshaped to 2 x 3:\n", reshaped)

# A simple matrix multiplication example
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A @ B =\n", A @ B)
```

#### 2.5 Aggregation and axis

```python
marks = np.array([
    [700, 12],
    [650, 8],
    [780, 15]
])

print("Column means:", marks.mean(axis=0))
print("Row sums:", marks.sum(axis=1))
print("Overall minimum:", marks.min())
print("Overall maximum:", marks.max())
```

#### 2.6 Practical tensor example – customer credit risk

```python
credit_risk_matrix = np.array([
    [25, 600, 720],
    [41, 900, 650],
    [33, 750, 780]
], dtype=float)

print("Credit risk feature matrix:\n", credit_risk_matrix)
print("Shape:", credit_risk_matrix.shape)
print("Number of customers:", credit_risk_matrix.shape[0])
print("Number of features:", credit_risk_matrix.shape[1])
print("Average credit score:", credit_risk_matrix[:, 2].mean())
```

**interpretation:** The matrix is a 2-D tensor with shape `(3, 3)`: three customers and three numerical features. NumPy makes it possible to calculate statistics and perform vectorised operations efficiently.

**conclusion:** Tensors are fundamental to numerical computing and machine learning. NumPy arrays provide a practical way to store and manipulate multi-dimensional numerical data.

## Part B – Data Acquisition

### Task 3 – Import datasets from multiple sources

### 3.1 Create a synthetic main CSV file

```python
main_data = pd.DataFrame({
    "customer_id": ["C001", "C002", "C003", "C004", "C005"],
    "age": [25, 41, 33, 52, 29],
    "annual_income": [600000, 900000, 750000, 1200000, 500000],
    "loan_amount": [150000, 400000, 250000, 500000, 180000],
    "loan_purpose": ["Education", "Home", "Car", "Business", "Other"],
    "credit_score": [720, 650, 780, 690, 710],
    "transaction_count": [45, 28, 60, 32, 40],
    "spending_ratio": [0.35, 0.62, 0.28, 0.55, 0.40],
    "default_flag": [0, 1, 0, 1, 0]
})

csv_path = DATA_DIR / "main_customer_loan_data.csv"
main_data.to_csv(csv_path, index=False)

print("CSV saved to:", csv_path.resolve())
display(main_data.head())
```

### 3.2 Read the CSV file

```python
csv_df = pd.read_csv(csv_path)
print("CSV shape:", csv_df.shape)
print("CSV columns:", csv_df.columns.tolist())
display(csv_df.head())
```

### 3.3 Create and parse a JSON customer metadata file

```python
customer_metadata = [
    {
        "customer_id": "C001", "gender": "Female", "region": "West",
        "education_level": "Graduate", "employment_type": "Salaried",
        "join_date": "2021-04-15"
    },
    {
        "customer_id": "C002", "gender": "Male", "region": "North",
        "education_level": "Post-Graduate", "employment_type": "Self-Employed",
        "join_date": "2019-08-20"
    },
    {
        "customer_id": "C003", "gender": "Female", "region": "South",
        "education_level": "Graduate", "employment_type": "Salaried",
        "join_date": "2022-01-10"
    },
    {
        "customer_id": "C004", "gender": "Male", "region": "East",
        "education_level": "Secondary", "employment_type": "Unemployed",
        "join_date": "2018-11-05"
    },
    {
        "customer_id": "C005", "gender": "Other", "region": "West",
        "education_level": "Primary", "employment_type": "Salaried",
        "join_date": "2023-06-12"
    }
]

json_path = DATA_DIR / "customer_metadata.json"
with open(json_path, "w", encoding="utf-8") as file:
    json.dump(customer_metadata, file, indent=4)

print("JSON saved to:", json_path.resolve())
```

```python
with open(json_path, "r", encoding="utf-8") as file:
    loaded_metadata = json.load(file)

json_df = pd.DataFrame(loaded_metadata)
print("JSON DataFrame shape:", json_df.shape)
display(json_df.head())
```

### 3.4 Create and query a SQL repayment-history table

```python
db_path = DATA_DIR / "loan_repayment_history.db"
connection = sqlite3.connect(db_path)

repayment_data = pd.DataFrame({
    "customer_id": ["C001", "C002", "C003", "C004", "C005"],
    "repayment_history": [0, 3, 0, 2, 1],
    "missed_payments_last_12m": [0, 3, 0, 2, 1]
})

repayment_data.to_sql(
    "repayment_history",
    connection,
    if_exists="replace",
    index=False
)

print("SQLite database created:", db_path.resolve())
```

```python
sql_query = '''
SELECT customer_id, repayment_history, missed_payments_last_12m
FROM repayment_history
ORDER BY customer_id;
'''

sql_df = pd.read_sql_query(sql_query, connection)
display(sql_df)
```

### 3.5 Fetch data from a dummy API

```python
def get_dummy_economic_api_response():
    # Simulated JSON response from an external economic-indicator API
    return {
        "source": "Dummy Economic Indicators API",
        "records": [
            {"region": "West", "unemployment_rate": 4.2, "inflation_rate": 5.1},
            {"region": "North", "unemployment_rate": 5.0, "inflation_rate": 5.3},
            {"region": "South", "unemployment_rate": 3.8, "inflation_rate": 4.9},
            {"region": "East", "unemployment_rate": 5.6, "inflation_rate": 5.5}
        ]
    }

api_response = get_dummy_economic_api_response()
api_df = pd.DataFrame(api_response["records"])

print("API source:", api_response["source"])
display(api_df)
```

### 3.6 Combine the data from all sources

```python
combined_df = (
    csv_df
    .merge(json_df, on="customer_id", how="left", validate="one_to_one")
    .merge(sql_df, on="customer_id", how="left", validate="one_to_one")
    .merge(api_df, on="region", how="left", validate="many_to_one")
)

print("Combined dataset shape:", combined_df.shape)
display(combined_df)
```

### 3.7 Basic acquisition validation

```python
print("Dataset shape:", combined_df.shape)
print("\nData types:")
print(combined_df.dtypes)

print("\nDuplicate customer IDs:", combined_df["customer_id"].duplicated().sum())

print("\nMissing values by column:")
print(combined_df.isna().sum())

print("\nTarget distribution:")
print(combined_df["default_flag"].value_counts(dropna=False))
```

### 3.8 Save the acquired dataset

```python
acquired_path = DATA_DIR / "acquired_customer_credit_risk.csv"
combined_df.to_csv(acquired_path, index=False)

print("Acquired dataset saved to:", acquired_path.resolve())

connection.close()
```
